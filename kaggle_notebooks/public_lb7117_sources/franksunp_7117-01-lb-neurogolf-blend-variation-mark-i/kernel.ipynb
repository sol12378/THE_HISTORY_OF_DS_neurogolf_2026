# [7117.01 LB] NeuroGolf Blend Variation Mark I

This version rebuilds a NeuroGolf ONNX submission from visible public notebook outputs and records a compact audit of the selected sources. The scored path is deterministic: attach Kokinn's Override Audit + Frank Delta notebook, validate each task against the available examples, keep the current public Mark I result as the B/reference line from the prior scored version, and write `/kaggle/working/submission.zip`.

Public upstream sources used here: Kokinn's `7117-01-lb-override-audit-frank-delta`, currently showing 7117.01, and the current public Mark I result, also showing 7117.01. The refreshed private A/B audit completed on Kaggle using Kokinn as A and Mark I as B. This public notebook keeps the executable source-visible path strict by attaching Kokinn directly and documenting the Mark I lineage without creating a circular notebook source. Prior Mark I, Seddik, Hairet, Biohack, Kokinn, NCTuan, Zoro, Kojimar, and other public runs remain useful context for the progression. Thanks to the upstream authors for publishing inspectable NeuroGolf artifacts.


```python
import csv
import hashlib
import json
import math
import os
import shutil
import subprocess
import sys
import tempfile
import time
import zipfile
from pathlib import Path

RUN_ID = "ngf_kokinn711701_marki711701_ab_audit_public_view_20260621_0135"
A_NAME = "A_kokinn_override_audit_frank_delta_7117_01"
B_NAME = "B_public_mark_i_v33_7117_01"
A_KEYWORDS = ["7117-01-lb-override-audit-frank-delta", "override-audit-frank-delta", "neurogolf-7114-70-self-task347-attrslice", "kokinnwakashuu"]
B_KEYWORDS = ["7117-01-lb-neurogolf-blend-variation-mark-i", "neurogolf-blend-variation-mark-i", "franksunp"]
WORK = Path("/kaggle/working")
INPUT = Path("/kaggle/input")
WORK.mkdir(parents=True, exist_ok=True)

for pkg in ["onnx", "onnxruntime"]:
    try:
        __import__(pkg)
    except Exception:
        subprocess.run([sys.executable, "-m", "pip", "install", "-q", pkg], check=False)

import numpy as np
import onnx
import onnxruntime as ort

for cand in [
    str(INPUT / "competitions" / "neurogolf-2026"),
    str(INPUT / "neurogolf-2026"),
    str(INPUT / "neurogolf_utils"),
]:
    if cand not in sys.path:
        sys.path.insert(0, cand)

try:
    import neurogolf_utils as ng
except Exception:
    ng = None


def log(event, **payload):
    row = {"time_utc": time.strftime("%Y-%m-%dT%H:%M:%SZ", time.gmtime()), "event": event}
    row.update(payload)
    with (WORK / "experiment_log.jsonl").open("a") as f:
        f.write(json.dumps(row, sort_keys=True) + "\n")
    print(json.dumps(row, sort_keys=True))


def find_task_dir():
    candidates = [INPUT / "competitions" / "neurogolf-2026", INPUT / "neurogolf-2026"]
    for p in candidates:
        if (p / "task001.json").exists():
            return p
    matches = list(INPUT.rglob("task001.json"))
    if not matches:
        raise FileNotFoundError("Could not locate NeuroGolf task JSON files")
    return matches[0].parent


TASK_DIR = find_task_dir()


def task_json(task_id):
    return json.loads((TASK_DIR / f"task{task_id:03d}.json").read_text())


def convert_to_numpy(example):
    if ng is not None and hasattr(ng, "convert_to_numpy"):
        return ng.convert_to_numpy(example)
    arrs = {}
    for mode in ["input", "output"]:
        grid = example[mode]
        if max(len(grid), len(grid[0])) > 30:
            return None
        arr = np.zeros((1, 10, 30, 30), dtype=np.float32)
        for r, row in enumerate(grid):
            for c, color in enumerate(row):
                arr[0, int(color), r, c] = 1.0
        arrs[mode] = arr
    return arrs


def sanitize_model(model):
    if ng is not None and hasattr(ng, "sanitize_model"):
        return ng.sanitize_model(model)
    return model


def calculate_params(model):
    if ng is not None and hasattr(ng, "calculate_params"):
        return ng.calculate_params(model)
    params = 0
    for init in model.graph.initializer:
        if any(dim <= 0 for dim in init.dims):
            return None
        params += math.prod(init.dims) if init.dims else 1
    for node in model.graph.node:
        if node.op_type != "Constant":
            continue
        for attr in node.attribute:
            if attr.name == "value":
                params += math.prod(attr.t.dims) if attr.t.dims else 1
            elif attr.name == "value_floats":
                params += len(attr.floats)
            elif attr.name == "value_ints":
                params += len(attr.ints)
    return int(params)


def calculate_memory(model, trace_path):
    if ng is not None and hasattr(ng, "calculate_memory"):
        return ng.calculate_memory(model, trace_path)
    onnx.checker.check_model(model, full_check=True)
    graph = onnx.shape_inference.infer_shapes(model, strict_mode=True).graph
    tensor_memory = {}
    tensor_dtypes = {}
    tensor_map = {t.name: t for t in list(graph.input) + list(graph.value_info) + list(graph.output)}
    for name, item in tensor_map.items():
        if name in {"input", "output"} or not item.type.HasField("tensor_type"):
            continue
        tensor_type = item.type.tensor_type
        if not tensor_type.HasField("shape"):
            return None
        num = 1
        for dim in tensor_type.shape.dim:
            if not dim.HasField("dim_value") or dim.dim_value <= 0:
                return None
            num *= int(dim.dim_value)
        dtype = onnx.helper.tensor_dtype_to_np_dtype(tensor_type.elem_type)
        tensor_memory[name] = num * np.dtype(dtype).itemsize
        tensor_dtypes[name] = dtype
    node_outputs = {node.name: list(node.output) for node in graph.node}
    with open(trace_path) as f:
        trace = json.load(f)
    for event in trace:
        if event.get("cat") != "Node" or "args" not in event:
            continue
        if "output_type_shape" not in event["args"]:
            continue
        node_name = event.get("name", "").replace("_kernel_time", "")
        if node_name not in node_outputs:
            continue
        for i, shape_dict in enumerate(event["args"]["output_type_shape"]):
            if i >= len(node_outputs[node_name]):
                continue
            out_name = node_outputs[node_name][i]
            if out_name not in tensor_dtypes:
                continue
            itemsize = np.dtype(tensor_dtypes[out_name]).itemsize
            mem = itemsize * sum(int(math.prod(dims)) for dims in shape_dict.values())
            tensor_memory[out_name] = max(tensor_memory.get(out_name, 0), mem)
    return int(sum(tensor_memory.values()))


def score_network(model, trace_path):
    if ng is not None and hasattr(ng, "score_network"):
        return ng.score_network(model, trace_path)
    return calculate_memory(model, trace_path), calculate_params(model)


def audit_one(model_path, examples):
    try:
        raw_model = onnx.load(str(model_path))
        sanitized = sanitize_model(raw_model)
        if sanitized is None:
            return {"status": "sanitize_fail", "n_fail": 10**9, "cost": None, "points": None, "ran": 0}
        tmpdir = Path(tempfile.mkdtemp())
        tmp = tmpdir / "model.onnx"
        onnx.save(sanitized, str(tmp))
        opts = ort.SessionOptions()
        opts.enable_profiling = True
        opts.graph_optimization_level = ort.GraphOptimizationLevel.ORT_DISABLE_ALL
        opts.log_severity_level = 3
        sess = ort.InferenceSession(str(tmp), opts, providers=["CPUExecutionProvider"])
        n_fail = 0
        ran = 0
        for ex in examples:
            pair = convert_to_numpy(ex)
            if pair is None:
                continue
            ran += 1
            pred = sess.run(["output"], {"input": pair["input"]})[0]
            if not np.array_equal((pred > 0.0).astype(np.float32), pair["output"]):
                n_fail += 1
        trace_path = sess.end_profiling()
        memory, params = score_network(sanitized, trace_path)
        try:
            os.remove(trace_path)
        except OSError:
            pass
        shutil.rmtree(tmpdir, ignore_errors=True)
        if memory is None or params is None:
            return {"status": "score_fail", "n_fail": n_fail, "cost": None, "points": None, "ran": ran}
        cost = int(memory + params)
        points = max(1.0, 25.0 - math.log(max(1.0, cost)))
        return {"status": "ok", "n_fail": n_fail, "cost": cost, "points": points, "ran": ran}
    except Exception as exc:
        return {
            "status": f"error:{type(exc).__name__}:{str(exc)[:180]}",
            "n_fail": 10**9,
            "cost": None,
            "points": None,
            "ran": 0,
        }


def extract_zip(path, dst):
    if dst.exists():
        shutil.rmtree(dst, ignore_errors=True)
    dst.mkdir(parents=True, exist_ok=True)
    with zipfile.ZipFile(path) as zf:
        zf.extractall(dst)
    roots = [p.parent for p in dst.rglob("task001.onnx")]
    return roots[0] if roots else dst


def text_matches(path, keywords):
    text = str(path).lower().replace("_", "-")
    return any(k.lower().replace("_", "-") in text for k in keywords)


def find_source_root(keywords, extract_dst, label, allow_any_zip_fallback=True):
    direct = []
    for p in INPUT.rglob("task001.onnx"):
        if text_matches(p, keywords):
            direct.append(p.parent)
    if direct:
        return sorted(direct, key=lambda x: len(str(x)))[0]

    zip_matches = []
    for z in INPUT.rglob("*.zip"):
        if text_matches(z, keywords) or text_matches(z.parent, keywords):
            zip_matches.append(z)
    if not zip_matches and allow_any_zip_fallback:
        # Fall back to any zip in a mounted notebook source; the exact mount names
        # can drift when Kaggle creates input aliases.
        zip_matches = [z for z in INPUT.rglob("submission.zip")]
    if len(zip_matches) == 1:
        return extract_zip(zip_matches[0], extract_dst)

    details = [str(z) for z in zip_matches[:20]]
    raise FileNotFoundError(f"Could not uniquely find {label} submission zip/task files: {details}")


def find_a_root():
    return find_source_root(A_KEYWORDS, Path("/tmp/a_kokinn_override_audit_frank_delta_711701"), "A Kokinn Override Audit Frank Delta 7117.01")


def find_b_root():
    return find_source_root(B_KEYWORDS, Path("/tmp/b_public_mark_i_v33_711701"), "B public Mark I v33 7117.01", allow_any_zip_fallback=False)


def task_path(root, task_id):
    name = f"task{task_id:03d}.onnx"
    direct = root / name
    if direct.exists():
        return direct
    matches = list(root.rglob(name))
    return matches[0] if matches else None


def better(a, b):
    aok = a["status"] == "ok" and a["n_fail"] == 0 and a["cost"] is not None
    bok = b["status"] == "ok" and b["n_fail"] == 0 and b["cost"] is not None
    if bok and not aok:
        return b
    if aok and not bok:
        return a
    if aok and bok and b["cost"] < a["cost"]:
        return b
    return a


def main():
    log("start", run_id=RUN_ID, task_dir=str(TASK_DIR), official_utils=bool(ng))
    a_root = find_a_root()
    try:
        b_root = find_b_root()
        b_available = True
    except FileNotFoundError as exc:
        b_root = None
        b_available = False
        log("b_source_unavailable", reason=str(exc))
    log("sources", A=str(a_root), B=str(b_root))

    examples_by_task = {}
    for task_id in range(1, 401):
        data = task_json(task_id)
        examples_by_task[task_id] = data.get("train", []) + data.get("test", []) + data.get("arc-gen", [])

    stage = WORK / "selected_submission"
    if stage.exists():
        shutil.rmtree(stage, ignore_errors=True)
    stage.mkdir(parents=True)

    rows = []
    picks = {A_NAME: 0, B_NAME: 0}
    for task_id in range(1, 401):
        a_path = task_path(a_root, task_id)
        b_path = task_path(b_root, task_id) if b_root is not None else None
        if a_path is None and b_path is None:
            raise RuntimeError(f"A and B missing task{task_id:03d}")
        examples = examples_by_task[task_id]
        if a_path is None:
            ra = {"status": "missing", "n_fail": 10**9, "cost": None, "points": None, "ran": 0}
        else:
            ra = audit_one(a_path, examples)
        if b_path is None:
            rb = {"status": "missing", "n_fail": 10**9, "cost": None, "points": None, "ran": 0}
        else:
            rb = audit_one(b_path, examples)
        ra.update({"source": A_NAME, "path": str(a_path) if a_path is not None else None})
        rb.update({"source": B_NAME, "path": str(b_path) if b_path is not None else None})
        winner = better(ra, rb)
        if not winner.get("path"):
            raise RuntimeError(f"No usable winner for task{task_id:03d}: A={ra['status']} B={rb['status']}")
        shutil.copy2(winner["path"], stage / f"task{task_id:03d}.onnx")
        picks[winner["source"]] += 1
        rows.append(
            {
                "task_id": f"task{task_id:03d}",
                "winner": winner["source"],
                "a_cost": ra.get("cost"),
                "b_cost": rb.get("cost"),
                "winner_cost": winner.get("cost"),
                "a_points": ra.get("points"),
                "b_points": rb.get("points"),
                "winner_points": winner.get("points"),
                "a_status": ra.get("status"),
                "b_status": rb.get("status"),
                "a_fail": ra.get("n_fail"),
                "b_fail": rb.get("n_fail"),
                "a_ran": ra.get("ran"),
                "b_ran": rb.get("ran"),
                "winner_path": winner.get("path"),
            }
        )
        if task_id % 25 == 0:
            log("audit_progress", task=task_id, picks=picks)

    picks_csv = WORK / "ab_audit_picks.csv"
    with picks_csv.open("w", newline="") as f:
        writer = csv.DictWriter(f, fieldnames=list(rows[0].keys()))
        writer.writeheader()
        writer.writerows(rows)

    submission_zip = WORK / "submission.zip"
    if submission_zip.exists():
        submission_zip.unlink()
    with zipfile.ZipFile(submission_zip, "w", compression=zipfile.ZIP_DEFLATED) as zf:
        for task_id in range(1, 401):
            zf.write(stage / f"task{task_id:03d}.onnx", arcname=f"task{task_id:03d}.onnx")

    total_points = sum(float(r["winner_points"] or 0.0) for r in rows)
    total_cost = sum(int(r["winner_cost"] or 0) for r in rows)
    a_points = sum(float(r["a_points"] or 0.0) for r in rows)
    b_points = sum(float(r["b_points"] or 0.0) for r in rows)
    manifest = {
        "run_id": RUN_ID,
        "approach": "A/B per-task audited cheapest valid selection with Kokinn Override Audit + Frank Delta 7117.01 as A and public Mark I v33 7117.01 as B/reference. Public notebook execution avoids circular self-source and falls back to A when B is not mounted.",
        "A": {
            "name": A_NAME,
            "score": 7117.01,
            "root": str(a_root),
            "notebook": "kokinnwakashuu/7117-01-lb-override-audit-frank-delta",
            "kernel_source": "kokinnwakashuu/7117-01-lb-override-audit-frank-delta",
        },
        "B": {
            "name": B_NAME,
            "score": 7117.01,
            "notebook": "franksunp/7117-01-lb-neurogolf-blend-variation-mark-i",
            "kernel_source": "franksunp/7117-01-lb-neurogolf-blend-variation-mark-i",
            "root": str(b_root) if b_root is not None else None,
            "available_in_runtime": b_available,
        },
        "picks": picks,
        "local_total_points": total_points,
        "local_total_cost": total_cost,
        "a_total_points": a_points,
        "b_total_points": b_points,
        "gain_vs_a": total_points - a_points,
        "gain_vs_b": total_points - b_points,
        "submission_sha256": hashlib.sha256(submission_zip.read_bytes()).hexdigest(),
        "submission_bytes": submission_zip.stat().st_size,
    }
    (WORK / "ab_audit_manifest.json").write_text(json.dumps(manifest, indent=2, sort_keys=True) + "\n")
    log("done", **manifest)


if __name__ == "__main__":
    main()


# Compact public audit summary.
import pandas as pd
import matplotlib.pyplot as plt
manifest = json.loads((WORK / 'ab_audit_manifest.json').read_text())
picks_df = pd.read_csv(WORK / 'ab_audit_picks.csv')
print('submission:', WORK / 'submission.zip')
print('submission_sha256:', manifest.get('submission_sha256'))
print('picks:', manifest.get('picks'))
print('local_total_points:', manifest.get('local_total_points'))
display(pd.DataFrame([
    {'role': 'A source', 'notebook': manifest['A']['notebook'], 'public_lb': manifest['A']['score']},
    {'role': 'B source', 'notebook': manifest['B']['notebook'], 'public_lb': manifest['B']['score']},
    {'role': 'previous Mark I context', 'notebook': 'franksunp/neurogolf-blend-variation-mark-i', 'public_lb': 7116.18},
]))
pick_counts = picks_df['winner'].value_counts().rename_axis('winner').reset_index(name='tasks')
display(pick_counts)
fig, ax = plt.subplots(figsize=(7, 4))
pick_counts.plot.bar(x='winner', y='tasks', ax=ax, legend=False, color=['#4C78A8', '#59A14F'])
ax.set_title('Selected tasks by public source')
ax.set_xlabel('')
ax.set_ylabel('tasks')
plt.tight_layout()
plt.show()
display(picks_df.sort_values('winner_cost', ascending=False).head(12))
```
