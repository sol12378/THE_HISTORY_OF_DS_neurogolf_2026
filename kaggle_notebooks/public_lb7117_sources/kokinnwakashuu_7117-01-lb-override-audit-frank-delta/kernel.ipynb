# Version 12 update: NeuroGolf 7117.01 override audit + Frank delta

This is the **7117.01 public LB** update of my earlier Graph Surgeon / self compact audit notebook. I am keeping the same Kaggle page history and votes attached while the artifact and audit trail move from the 7116.91 override family to the first confirmed 7117+ public artifact in this line.

Version 11 moved the base to Kojimar / islet's public **[7116.91 LB] Audited NeuroGolf ONNX Overrides** and re-applied audited `graph.value_info` cleanup. Version 12 adds two task-level deltas found by comparing Frank's 7116.79 blend artifact against the 7116.91 compact base.

The adopted changes are Frank's `task349` and `task364` models. Each was probed as a single-task submission first, and then the pair was submitted together. The combined artifact scored **7117.01** public LB and is emitted here as an exact base64-embedded zip.

Thanks to the upstream public notebook authors, especially Seddik for the Graph Surgeon family, Kojimar / islet for the 7116.91 override release, and Frank for the blend artifact that exposed the two useful deltas.

## What changed from 7116.66 to 7116.91

Kojimar's 7116.91 notebook is intentionally small: it merges a public base submission with six ONNX overrides.

| Task group | Change | Audit note |
|---|---|---|
| `044, 105` | override versions replace my previous compact versions | part of the 7116.91 public artifact |
| `233, 255, 286, 347` | new task-level overrides | net-new public improvement over the 7116.66 family |
| Version 10 compact tasks | 20 old `value_info` strips were absent from the 7116.91 asset bundle | re-applied after audit where safe |

Diffing the rebuilt 7116.91 zip against the Version 10 artifact produced 24 changed files. Twenty of those were simply the old compact tasks becoming larger again, while the real new public score movement came from the override family, especially the net-new tasks above.

## Frank 7116.79 delta scan

I also downloaded Frank's public **[7116.79 LB] NeuroGolf Blend Variation Mark I** output and compared its `selected_submission` task-by-task against the 7116.91 compact artifact.

Most smaller files were not useful: several were older versions of tasks already improved by the 7116.91 override bundle, and some local-size wins had equal or worse official-style local cost. Two tasks were different:

| Task | Source | Local effect | Single-task public probe | Decision |
|---:|---|---:|---:|---|
| 349 | Frank selected submission | cost improves by 2699 | 7116.97 | accept |
| 364 | Frank selected submission | cost improves by 1320 | 7116.94 | accept |
| 349 + 364 | combined on top of 7116.91 compact | combined local gain 4019 | 7117.01 | current artifact |

This is the kind of public artifact blend that is easy to overfit, so I did not adopt the whole Frank bundle. I only took the two task-level replacements that passed single-task public probes and then survived the combined submission.
